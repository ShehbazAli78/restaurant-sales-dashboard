```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df["year"] = df["order_date"].dt.year
df["month"] = df["order_date"].dt.month
df.groupby("month")["order_id"].count()
```

```python
df.groupby("month")["order_id"].count().plot(kind="line")
plt.show()
```
